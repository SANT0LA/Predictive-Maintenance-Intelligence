```python
df['soma_subtipos'] = df['TWF'] + df['HDF'] + df['PWF']+ df['OSF'] + df['RNF']
df['soma_subtipos']
```

```python
caso1 = df[(df['Machine failure'] == 1) & (df['soma_subtipos'] == 0)]
print(caso1.shape[0])
```

```python
df_clean = df[~((df['Machine failure'] == 1) & (df['soma_subtipos'] == 0)) & 
              ~((df['Machine failure'] == 0) & (df['soma_subtipos'] >= 1))]

print(df_clean.shape)
```
